```python
from ImageScience import ImageScience
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import astropy.units as u
from Functions import *
import glob
from dust_extinction.parameter_averages import CCM89

Rv = 3.1
ext = CCM89(Rv=Rv)

ha = 6562.8 * u.AA
paa = 1.8756 * u.um

k_ha = Rv * ext(ha)
k_paa = Rv * ext(paa)


def get_ebv_from_paalpha_halpha(
    pa_alpha_flux,
    h_alpha_flux,
    intrinsic_ratio=0.122,
    k_halpha=2.5351102403705306,
    k_paalpha=0.4549747533172519,
):
    """
    Compute E(B-V) from Pa-alpha / H-alpha.

    Parameters
    ----------
    pa_alpha_flux : float or ndarray

    h_alpha_flux : float or ndarray

    intrinsic_ratio : float
        Intrinsic Case B ratio (Paα/Hα).

    k_halpha : float
        Extinction curve value at Hα.

    k_paalpha : float
        Extinction curve value at Paα.

    Returns
    -------
    ebv : float or ndarray
    """

    ratio = pa_alpha_flux / h_alpha_flux

    ebv = (
        2.5
        * np.log10(ratio / intrinsic_ratio)
        / (k_halpha - k_paalpha)
    )

    return ebv


def flux_to_abmag(flux):
    """
    Convert spectral flux density to an AB magnitude.

    Parameters
    ----------
    flux : astropy.units.Quantity
        Spectral flux density with units equivalent to
        W/m^2/Hz (or Jy, erg/s/cm^2/Hz, etc.).

    Returns
    -------
    mag : float or ndarray
        AB magnitude.
    """

    flux = flux.to(u.Jy)

    mag = -2.5 * np.log10(flux.value / 3631.0)

    return mag
```
